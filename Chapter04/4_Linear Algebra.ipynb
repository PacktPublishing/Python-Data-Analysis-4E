## Working with Scalar, Vector, Matrix, and Tensor

```python
import numpy as np

# Scalar - 0-Dimension - Single element
scalar = np.array(5) 
print("Scalar:", scalar, "Shape:", scalar.shape)

# Vector - 1-Dimension
vector = np.array([1, 2, 3])
print("Vector:", vector, "Shape:", vector.shape)

# Matrix - 2-Dimension
matrix = np.array([[1, 2], [3, 4]])
print("Matrix:\n", matrix, "Shape:", matrix.shape)

# Tensor - 3-Dimension
tensor = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print("Tensor:\n", tensor, "Shape:", tensor.shape)
```

```python
# Create two vectors
v1 = np.array([1, 2])
v2 = np.array([3, 4])

# Dot product
print("Dot Product:", np.dot(v1, v2))

# Magnitude
print("Norm:", np.linalg.norm(v1))
```

```python
# Create two matrix
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Matrix multiplication
print("Matrix Multiplication:\n", np.dot(A, B))

# Transpose
print("Transpose:\n", A.T)
```

## Fitting polynomials with NumPy 

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate two random vectors
v1 = np.random.rand(10)
v2 = np.random.rand(10)

# Step 2: Create a sequence of equally spaced values within the range of v1
sequence = np.linspace(v1.min(), v1.max(), num=len(v1) * 10)

# Step 3: Fit a polynomial of degree 3 (cubic) to the data
degree = 3
coefs = np.polyfit(v1, v2, degree)

# Step 4: Evaluate the polynomial at the values in the sequence
polynomial_values = np.polyval(coefs, sequence)

# Step 5: Plot the polynomial curve
plt.figure(figsize=(8, 5))
plt.plot(sequence, polynomial_values, label=f'Polynomial Fit (Degree {degree})')
plt.scatter(v1, v2, color='red', label='Original Data')
plt.title('Polynomial Curve Fitting using NumPy')
plt.xlabel('v1')
plt.ylabel('v2')
plt.legend()
plt.grid(True)
plt.tight_layout()

# Step 6: Show the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate two random vectors (simulated data)
v1 = np.random.rand(10)
v2 = np.random.rand(10)

# Step 2: Create a smooth sequence of x-values for plotting curves
sequence = np.linspace(v1.min(), v1.max(), num=200)

# Step 3: Define polynomial degrees to compare
degrees = [3, 5, 7]

# Step 4: Plot original data
plt.figure(figsize=(10, 6))
plt.scatter(v1, v2, color='black', label='Original Data', zorder=5)

# Step 5: Fit and plot polynomial curves for each degree
for deg in degrees:
    coefs = np.polyfit(v1, v2, deg)
    poly_vals = np.polyval(coefs, sequence)
    plt.plot(sequence, poly_vals, label=f'Degree {deg}')

# Step 6: Customize plot
plt.title('Polynomial Curve Fitting (Degrees 3, 5, 7)')
plt.xlabel('v1')
plt.ylabel('v2')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Determinant

```python
# Import numpy 
import numpy as np

# Create matrix using NumPy
mat = np.mat([[2, 4], [5, 7]])
print("Matrix:\n", mat)

# Calculate determinant
det = np.linalg.det(mat)
print("Determinant:", round(det, 2))
```

## Matrix inverse using NumPy

## Solving linear equations using NumPy 

```python
# Create matrix A and Vector B using NumPy
A=np.mat([[1,1],
          [3,2]])
print("Matrix A:\n",A)

B = np.array([200,450]) 
print("Vector B:", B)

# Solve linear equations
solution = np.linalg.solve(A, B) 
print("Solution vector x:", solution)

# Check the solution
print("Result:",np.dot(A,solution))
```

## Decomposing a matrix using SVD 

```python
# Import required libraries
import numpy as np
from scipy.linalg import svd

# Step 1: Create the matrix
matrix = np.array([[5, 3, 1],
                   [5, 3, 0],
                   [1, 0, 5]])

# Step 2: Perform Singular Value Decomposition
U, sigma, Vt = svd(matrix)

# Step 3: Display the results
print("Left Singular Matrix (U):\n", np.round(U, 2))
print("\nSingular Values (Sigma):\n", np.round(sigma, 2))
print("\nRight Singular Matrix (V^T):\n", np.round(Vt, 2))
```

## LU DEcomposition 

```python
import numpy as np
from scipy.linalg import lu

# Create 3x3 matrix A
A = np.array([
    [1, 3, 4], 
    [5, 6, 9], 
    [1, 2, 3]]
)

# Perform LU decomposition
P, L, U = lu(A)

# Print components
print("Original Matrix A:\n", A)
print("Permutation Matrix P:\n", P)
print("Lower Triangular Matrix L:\n", L)
print("Upper Triangular Matrix U:\n", U)
```

## QR Decomposition 

```python
import numpy as np

# Create 3x4 matrix A
A = np.array([
    [1, 3, 4], 
    [5, 6, 9], 
    [1, 2, 3], 
    [7, 6, 8]]
)

# Perform QR Decomposition using numpy.linalg module
Q, R = np.linalg.qr(A)

# print outputs
print("Original Matrix A:\n", A)
print("\nOrthogonal Matrix Q:\n", np.round(Q, 2))
print("\nUpper Triangular Matrix R:\n", np.round(R, 2))

# Recreate the Original matrix
A_original = Q @ R
print("\nOriginal Input Matrix A:\n", np.round(A_original, 2))
```

## Binomial Distribution

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt

# Create an numpy vector of size 5000 with value 0
cash_balance = np.zeros(5000) 

cash_balance[0] = 500

# Generate random numbers using Binomial
samples = np.random.binomial(9, 0.5, size=len(cash_balance))

# Update the cash balance
for i in range(1, len(cash_balance)):
    if samples[i] < 5:
        cash_balance[i] = cash_balance[i - 1] - 1 
    else:
        cash_balance[i] = cash_balance[i - 1] + 1 

# Plot the updated cash balance
plt.plot(np.arange(len(cash_balance)), cash_balance)
plt.show()
```

## Log-Normal Distribution

```python
import numpy as np 
import matplotlib.pyplot as plt 

# Mean, Standard Deviation, and Size of the distribution  
mean, std_dev, size = 10, 2, 10000 

# Create data based on normal distribution 
data = np.random.lognormal(mean, std_dev, size)   

# Create histogram plot  
plt.hist(data,  
         bins=100,  
         alpha=0.6,  
         color='blue') 

# Set x-axis limit 
plt.xlim(0, 6000000) 
plt.xlabel('Lognormal Distribution') 
plt.show() 
```

## Creating a masked array using numpy.ma subpackage

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from skimage.data import astronaut 
from skimage.util import img_as_float

# Load image and convert to float for log transform
face_image = img_as_float(astronaut())

# Generate a random mask: 0 = unmasked, 1 = masked
mask_random_array = np.random.randint(0, 2, size=face_image.shape, dtype=bool)

# Create masked array
masked_image = np.ma.array(face_image, mask=mask_random_array)

# Apply Log operation on original image
log_original = np.log1p(face_image)      
log_masked   = np.log1p(masked_image)  

# Plot all subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
titles = ["Original Image", "Masked Image", "Log Transform on Original", "Log Transform on Masked"]
images = [face_image, masked_image, log_original, log_masked]

# Display all four images
for ax, img, title in zip(axes.flat, images, titles):
    ax.imshow(img)
    ax.set_title(title)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

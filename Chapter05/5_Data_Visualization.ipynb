```python
# import essential library matplotlib
import matplotlib.pyplot as plt
 
# create the data
x = [1,3,5,7,9,11]
y = [10,25,35,33,41,59]
 
# Draw the scatter chart
plt.scatter(x, y, c='red', marker='o', s=80, alpha=0.7, label='Data Points')
 
# Append the label on X-axis
plt.xlabel("X-label")
 
# Append the label on X-axis
plt.ylabel("Y-label")


# annotate some points
for i in range(len(x)):
    plt.annotate(f"({x[i]}, {y[i]})", (x[i], y[i]), textcoords="offset points", xytext=(5,5))

# Add the title to graph
plt.title("Scatter Chart Sample")
 
# Add grid and legend
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Display the chart
plt.show()
```

## Bubble Plot

```python
# Import the required modules
import matplotlib.pyplot as plt
import numpy as np

# Set figure size
plt.figure(figsize=(11, 6))

# Create the data
countries = ['Qatar', 'Luxembourg', 'Singapore', 'Norway', 'UAE',
             'USA', 'Brazil', 'India', 'Kenya']

# Population in millions
populations = [2.8, 0.65, 5.9, 5.5, 9.9, 335, 216, 1428, 55]

# GDP per capita in USD
gdp_per_capita = [130000, 125000, 90000, 88000, 75000,
                  76000, 9500, 2500, 2100]

# Convert GDP list to NumPy array
gdp_array = np.array(gdp_per_capita)

# Scale GDP per capita for better bubble size visibility
bubble_sizes = 100 + ((gdp_array - gdp_array.min()) /
                      (gdp_array.max() - gdp_array.min())) * 1200

# Draw the bubble chart
plt.scatter(
    countries,
    populations,
    s=bubble_sizes,
    color="lightblue",
    edgecolors="black",
    alpha=0.7
)

# Add X and Y labels
plt.xlabel("Countries", fontsize=12)
plt.ylabel("Population in Millions", fontsize=12)

# Add title
plt.title("Country Population vs GDP per Capita (Bubble Chart)",
          fontsize=14, fontweight="bold")

# Rotate x labels
plt.xticks(rotation=45)

# Add grid
plt.grid(True, linestyle="--", alpha=0.6)

# Function to create bubble size for legend
def size_from_gdp(value):
    return 100 + ((value - gdp_array.min()) /
                  (gdp_array.max() - gdp_array.min())) * 1200

# Add bubble size legend
legend_values = [5000, 50000, 100000, 130000]

legend_handles = [
    plt.scatter([], [], s=size_from_gdp(value),
                color="lightblue", edgecolors="black", alpha=0.7,
                label=f"${value:,}")
    for value in legend_values
]

plt.legend(
    handles=legend_handles,
    title="GDP per Capita\n(Bubble Size)",
    loc="upper right",
    frameon=True
)

# Improve layout
plt.tight_layout()

# Show the plot
plt.show()
```

## Subplots

```python
# Import library
import matplotlib.pyplot as plt
import numpy as np

# Create some sample data
x = np.linspace(0, 10, 50)
y = np.sin(x)
z = np.cos(x)

# Create a 2x2 subplot grid (2 rows, 2 columns)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# -Plot 1: Line chart 
axes[0, 0].plot(x, y, color='blue', label="sin(x)")
axes[0, 0].set_title("Line Plot")
axes[0, 0].legend()
axes[0, 0].grid(True, linestyle="--", alpha=0.5)

# Plot 2: Bar chart
# Create the data
movie_ratings = [1, 2, 3, 4, 5, 6]
rating_counts = [21, 45, 72, 89, 42, 30]

# Plot the bar chart
axes[0, 1].bar(movie_ratings, rating_counts, color="blue", edgecolor="black")
axes[0, 1].set_title("Bar Chart")

# Plot 3: Scatter plot
axes[1, 0].scatter(x, z, c='red', marker='o', alpha=0.6)
axes[1, 0].set_title("Scatter Plot")

# Plot 4: Pie
subjects = ["Mathematics", "Science", "Communication Skills", "Computer Application", "History"]
scores = [85, 62, 57, 92, 70]

axes[1, 1].pie(scores,
        labels=subjects,
        colors=['red','green','blue','yellow','skyblue'],
        startangle=90,
        shadow=True,
        explode=(0,0.1,0,0,0),
        autopct='%1.1f%%',
        wedgeprops={'edgecolor':'black'})
axes[1, 1].set_title("Pie Chart")

# Adjust layout
fig.suptitle("Multiple Subplots Example", fontsize=16, fontweight='bold')
plt.tight_layout(rect=[0, 0, 1, 0.96])

# Show the plot
plt.show()
```

# Visualization using Pandas

```python
# Import required modules
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# Create the DataFrame
df = pd.DataFrame({
    'name': ['Ajay','Malala','Abhijeet','Yming','Desilva','Lisa'],
    'age': [22,72,25,19,42,38],
    'gender': ['M','F','M','M','M','F'],
    'country': ['India','Pakistan','Bangladesh','China','Srilanka','UK'],
    'income': [2500,3800,3300,2100,4500,5500]
})

# Map genders to colors
color_map = {'M': 'blue', 'F': 'red'}
df['color'] = df['gender'].map(color_map)

# Create scatter plot using pandas plot
ax = df.plot(
    kind='scatter',
    x='age',
    y='income',
    c=df['color'],
    s=120,
    edgecolor="black",
    alpha=0.7,
    figsize=(8, 6),
    title="Age vs Income by Gender"
)

# Add axis labels
ax.set_xlabel("Age", fontsize=12)
ax.set_ylabel("Income", fontsize=12)

# Annotate names above points
for i, row in df.iterrows():
    ax.text(row['age'] + 0.5, row['income'] + 50, row['name'], fontsize=9)

# Add grid
ax.grid(True, linestyle="--", alpha=0.6)

# Add legend for gender colors
legend_elements = [
    Line2D([0], [0], marker='o', color='w', label='Male',
           markerfacecolor='blue', markeredgecolor='black', markersize=10),
    Line2D([0], [0], marker='o', color='w', label='Female',
           markerfacecolor='red', markeredgecolor='black', markersize=10)
]

ax.legend(handles=legend_elements, title="Gender")

# Show plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.patches import Patch

# Sample DataFrame
df = pd.DataFrame({
    'name': ['Ajay','Malala','Abhijeet','Yming','Desilva','Lisa'],
    'age': [22,72,25,19,42,38],
    'gender': ['M','F','M','M','M','F'],
})

# Map genders to colors
color_map = {'M':'blue', 'F':'red'}
df['color'] = df['gender'].map(color_map)

# Create bar plot using pandas plot
ax = df.plot(
    kind='bar',
    x='name',
    y='age',
    figsize=(8, 6),
    color=df['color'],
    legend=False,
    title="Age of Individuals"
)

# Add axis labels
ax.set_xlabel("Name", fontsize=12)
ax.set_ylabel("Age", fontsize=12)

# Rotate x-axis labels for clarity
plt.xticks(rotation=30, ha='right')

# Add grid lines
ax.grid(axis='y', linestyle="--", alpha=0.6)

# Add legend for gender colors
legend_elements = [
    Patch(facecolor='blue', label='Male'),
    Patch(facecolor='red', label='Female')
]

ax.legend(handles=legend_elements, title="Gender")

# Annotate bar values
for p in ax.patches:
    ax.annotate(
        str(p.get_height()),
        (p.get_x() + p.get_width() / 2, p.get_height() + 1),
        ha='center',
        fontsize=10
    )

# Show plot
plt.show()
```

```python
# Import the required libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create DataFrame
df = pd.DataFrame(
    {
        'x': [1, 3, 5, 7, 9, 11],
        'y': [10, 25, 35, 33, 41, 59]
    }
)

# Set a theme style
sns.set_theme(style="whitegrid")

# Create lmplot
plot = sns.lmplot(
    x='x',
    y='y',
    data=df,
    height=6,        
    aspect=1.5,     
    ci=95,              
    scatter_kws={"s": 70, "color": "blue", "alpha": 0.7},   # scatter point size and color
    line_kws={"color": "red", "lw": 2}                      # regression line color and width
)

# Add labels and title
plot.set_axis_labels("X values", "Y values")
plt.title("Linear Regression Fit", fontsize=14, fontweight="bold")

# Show plot
plt.show()
```

## Gantt Chart

```python
import plotly.express as px
import pandas as pd

# Create Dataframe
df = pd.DataFrame([
    dict(Trip="Trip 1", Start="2025-09-14 08:00", Finish="2025-09-14 10:30", Driver="Joy",  Vehicle="V-101"),
    dict(Trip="Trip 2", Start="2025-09-14 11:00", Finish="2025-09-14 12:00", Driver="Joy",  Vehicle="V-101"), 
    dict(Trip="Trip 3", Start="2025-09-14 09:15", Finish="2025-09-14 11:45", Driver="Adam",   Vehicle="V-203"),
    dict(Trip="Trip 4", Start="2025-09-14 12:30", Finish="2025-09-14 14:00", Driver="Adam",   Vehicle="V-203"),
    dict(Trip="Trip 5", Start="2025-09-14 08:30", Finish="2025-09-14 12:30", Driver="Josh", Vehicle="V-307"),
    dict(Trip="Trip 6", Start="2025-09-14 13:00", Finish="2025-09-14 16:00", Driver="Josh", Vehicle="V-307"),
])

# Create Gantt Chart 
fig = px.timeline(
    df, x_start="Start", x_end="Finish", y="Trip",
    color="Driver",
    hover_data=["Vehicle",]
)

# Show most recent task at the top 
fig.update_yaxes(autorange="reversed")

# Set Labels and bar gap
fig.update_layout(
    title="Vehicle Trips Schedule",
    bargap=0.3,
    xaxis_title="Time",
    yaxis_title="Trips",
    xaxis=dict(tickformat="%H:%M\n%d %b %Y")
)

fig.show()
```

## subplots using plotly

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Sample data
x = np.linspace(0, 10, 50)
y = np.sin(x)
z = np.cos(x)

# Data for bar chart
movie_ratings = [1, 2, 3, 4, 5, 6]
rating_counts = [21, 45, 72, 89, 42, 30]

# Data for pie chart
subjects = ["Mathematics", "Science", "Communication Skills", "Computer Application", "History"]
scores = [85, 62, 57, 92, 70]

# Create 2x2 subplot grid
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=("Line Plot", "Bar Chart", "Scatter Plot", "Pie Chart"),
    specs=[[{"type": "xy"}, {"type": "xy"}],
           [{"type": "xy"}, {"type": "domain"}]]
)

# Plot 1: Line chart
fig.add_trace(
    go.Scatter(x=x, y=y, mode="lines", name="sin(x)", line=dict(color="blue")),
    row=1, col=1
)

# Plot 2: Bar chart
fig.add_trace(
    go.Bar(x=movie_ratings, y=rating_counts, marker=dict(color="blue", line=dict(color="black", width=1))),
    row=1, col=2
)

# Plot 3: Scatter plot
fig.add_trace(
    go.Scatter(x=x, y=z, mode="markers", marker=dict(color="red", size=8, opacity=0.6), name="cos(x)"),
    row=2, col=1
)

# Plot 4: Pie chart
fig.add_trace(
    go.Pie(labels=subjects, values=scores,
           hole=0,  # full pie
           marker=dict(colors=['red','green','blue','yellow','skyblue'], 
                       line=dict(color="black", width=1)),
           pull=[0,0.1,0,0,0]),
    row=2, col=2
)

# Layout - title, size and legend
fig.update_layout(
    title_text="Plotly Subplots Example",
    height=800, width=900,
    showlegend=True
)

fig.show()
```

## Custom Buttons

```python
# Import graph_objects
import plotly.graph_objects as go

# Sample data
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug", "Sep","Oct","Nov","Dec"]
actual_sales = [120, 135, 150, 160, 170, 180, 190, 210, 200, 190, 230, 250]
forecasted_sales = [115, 140, 145, 155, 165, 185, 195, 205, 200, 210, 220, 230]

# Create figure object with two lines
fig = go.Figure([
    go.Scatter(x=months, y=actual_sales, mode="lines+markers", 
                         name="Actual Sales", line=dict(color="blue")),
    go.Scatter(x=months, y=forecasted_sales, mode="lines", 
                         name="Predicted Sales", line=dict(color="red", dash="dash"))
])

# Layout - title, and buttons
fig.update_layout(
    title="Custom Buttons Example",
    updatemenus=[dict(
        type="buttons",
        direction="right",
        x=0.5, y=1.15, xanchor="center",
        buttons=[ # Add 3 buttons
            dict(label="Actual Sales",
                 method="update",
                 args=[{"visible":[True, False]}]), # Actual 
            dict(label="Forecasted Sales",
                 method="update",
                 args=[{"visible":[False, True]}]), # Forecast
            dict(label="Show Both Sales",
                 method="update",
                 args=[{"visible":[True, True]}]), # Both Sales
        ]
    )]
)

# Display plot
fig.show()
```

## Dropdowns

```python
import pandas as pd
import plotly.graph_objects as go

# Load the dataset
df=pd.read_csv("HR_comma_sep.csv")

# Define colors - blue for stayed, red for left
left_color = {0: "skyblue", 1: "red"}  
salaries = ["low", "medium", "high"] 

fig = go.Figure()

# Create traces like (low,0), (low,1), (medium,0), (medium,1), (high,0), (high,1)
trace_tags = []
for sal in salaries:
    for l in [0, 1]:
        m = (df["salary"] == sal) & (df["left"] == l)
        fig.add_trace(go.Scatter(
            x=df.loc[m, "satisfaction_level"],
            y=df.loc[m, "last_evaluation"],
            mode="markers",
            name=f"{sal.title()} | left={l}",
            marker=dict(color=left_color[l], size=7, opacity=0.75, line=dict(width=0.5,color="rgba(0,0,0,0.4)")),
            hovertemplate=("salary: "+sal+
                           "<br>left: "+str(l)+
                           "<br>satisfaction: %{x}<br>evaluation: %{y}<extra></extra>")
        ))
        trace_tags.append((sal, l))

# Create visualization mask
def vis_mask(show_salaries):
    return [ (sal in show_salaries) for (sal, l) in trace_tags ]

# Layout - title, labels and buttons
fig.update_layout(
    title="Employee Churn Analysis — filter by Salary",
    template="plotly_white",
    xaxis_title="Satisfaction Level",
    yaxis_title="Performance",
    updatemenus=[
        dict(
            type="dropdown", x=1.05, y=0.6, xanchor="left", showactive=True,
            buttons=[
                dict(label="All salaries",   method="update",
                     args=[{"visible": vis_mask({"low","medium","high"})},
                           {"title": "Employee Churn Analysis — All salaries"}]),
                dict(label="Low",            method="update",
                     args=[{"visible": vis_mask({"low"})},
                           {"title": "Employee Churn Analysis — Low salary"}]),
                dict(label="Medium",         method="update",
                     args=[{"visible": vis_mask({"medium"})},
                           {"title": "Employee Churn Analysis — Medium salary"}]),
                dict(label="High",           method="update",
                     args=[{"visible": vis_mask({"high"})},
                           {"title": "Employee Churn Analysis — High salary"}]),
            ]
        )
    ],
    legend_title_text="Salary | left",
    hovermode="closest",
    height=600
)

fig.show()
```

## sliders

```python
import plotly.graph_objects as go
import pandas as pd

# Create a DataFrame for sales data
sales = {
    "Year": ["2018", "2019", "2020", "2021", "2022"],
    "Laptop": [120, 180, 220, 280, 320],
    "Smartphone": [150, 120, 300, 350, 380],
    "Tablet": [200, 250, 150, 400, 420]
}
df = pd.DataFrame(sales)

products = ["Laptop", "Smartphone", "Tablet"]
years = df["Year"].tolist()

# Create traces
traces = []
for yr in years:
    row = df[df["Year"] == yr]
    traces.append(go.Bar(
        x=products,
        y=row[products].values.flatten(),  # get product sales values
        name=yr,
        visible=(yr == "2018"),  # show only 2018 initially
        marker=dict(line=dict(color="black", width=1))
    ))

# Create slider steps
steps = []
for i, yr in enumerate(years):
    steps.append(dict(
        method="update",
        args=[
            {"visible": [j == i for j in range(len(traces))]},
            {"title": f"Product Sales in {yr}"}
        ],
        label=yr
    ))

# Layout - title, labels and sliders
fig = go.Figure(traces)
fig.update_layout(
    title="Product Sales by Year",
    xaxis_title="Products",
    yaxis_title="Units Sold",
    template="plotly_white",
    sliders=[dict(active=0, steps=steps)]
)

fig.show()
```

# Building Analytical Dashboards with Dash Framework

## First Interactive plot in Dash

```python
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.express as px
import pandas as pd

# Sample data
df = pd.DataFrame({
    "Year": ["2018", "2019", "2020", "2021", "2022"],
    "Laptop": [120, 180, 220, 280, 320],
    "Smartphone": [150, 120, 300, 350, 380],
    "Tablet": [200, 250, 150, 400, 420]
})

# Create Dash app
app = dash.Dash(__name__)

# Layout
app.layout = html.Div([
    html.H1("Product Sales Dashboard"),

    dcc.Dropdown(
        id='year-dropdown',
        options=[{'label': y, 'value': y} for y in df['Year']],
        value="2018"
    ),

    dcc.Graph(id='sales-graph')
])

# Callback function
@app.callback(
    Output('sales-graph', 'figure'),
    Input('year-dropdown', 'value')
)

def update_graph(selected_year):

    filtered_df = df[df["Year"] == selected_year]

    # Convert row into long format for plotting
    data = pd.DataFrame({
        "Product": ["Laptop", "Smartphone", "Tablet"],
        "Sales": [
            filtered_df["Laptop"].values[0],
            filtered_df["Smartphone"].values[0],
            filtered_df["Tablet"].values[0]
        ]
    })

    fig = px.bar(
        data,
        x="Product",
        y="Sales",
        title=f"Product Sales in {selected_year}",
        color="Product"
    )

    return fig

if __name__ == '__main__':
    app.run(debug=True)
```

## Multi-Page Layout in Dash

```python
import dash
from dash import dcc, html, dash_table
from dash.dependencies import Input, Output
import plotly.express as px
import pandas as pd

# Sample data
df = pd.DataFrame({
    "Year": ["2018", "2019", "2020", "2021", "2022"],
    "Laptop": [120, 180, 220, 280, 320],
    "Smartphone": [150, 120, 300, 350, 380],
    "Tablet": [200, 250, 150, 400, 420]
})

# Create Dash app
app = dash.Dash(__name__)

# Layout
app.layout = html.Div([
    html.H1("Product Sales Dashboard"),

    dcc.Tabs(id="tabs", value="tab-chart", children=[
        dcc.Tab(label="Sales Chart", value="tab-chart"),
        dcc.Tab(label="Sales Table", value="tab-table"),
    ]),

    html.Div(id="tab-content")
])

# Callback to render tab content
@app.callback(
    Output("tab-content", "children"),
    Input("tabs", "value")
)
def render_tab_content(tab):
    if tab == "tab-chart":
        return html.Div([
            html.Br(),
            dcc.Dropdown(
                id="year-dropdown",
                options=[{"label": y, "value": y} for y in df["Year"]],
                value="2018",
                clearable=False,
                style={"width": "50%"}
            ),
            html.Br(),
            dcc.Graph(id="sales-graph")
        ])

    elif tab == "tab-table":
        return html.Div([
            html.Br(),
            html.H3("Product Sales Table"),
            dash_table.DataTable(
                data=df.to_dict("records"),
                columns=[{"name": i, "id": i} for i in df.columns],
                page_size=5,
                sort_action="native",
                filter_action="native",
                style_table={"overflowX": "auto"},
                style_cell={"textAlign": "center"},
                style_header={"fontWeight": "bold"}
            )
        ])

# Callback to update chart
@app.callback(
    Output("sales-graph", "figure"),
    Input("year-dropdown", "value")
)
def update_graph(selected_year):
    filtered_df = df[df["Year"] == selected_year]

    chart_df = pd.DataFrame({
        "Product": ["Laptop", "Smartphone", "Tablet"],
        "Sales": [
            filtered_df["Laptop"].values[0],
            filtered_df["Smartphone"].values[0],
            filtered_df["Tablet"].values[0]
        ]
    })

    fig = px.bar(
        chart_df,
        x="Product",
        y="Sales",
        color="Product",
        title=f"Product Sales in {selected_year}"
    )
    return fig

if __name__ == "__main__":
    app.run(debug=True)
```

## Real-Time Updates

```python
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.express as px
import pandas as pd
import random

app = dash.Dash(__name__)

temperature = []

app.layout = html.Div([
    html.H2("Live Temperature Sensor"),

    dcc.Graph(id='live-temp'),

    dcc.Interval(
        id='interval',
        interval=2000,
        n_intervals=0
    )
])

@app.callback(
    Output('live-temp','figure'),
    Input('interval','n_intervals')
)
def update_graph(n):

    temperature.append(random.randint(20,35))

    data = pd.DataFrame({
        "Time": list(range(len(temperature))),
        "Temperature": temperature
    })

    fig = px.line(data, x="Time", y="Temperature",
                  title="Temperature Sensor Monitoring")

    return fig

if __name__ == '__main__':
    app.run(debug=True)
```

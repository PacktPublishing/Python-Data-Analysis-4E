## Creating NumPy views and copies

```python
# Create NumPy Array
arr = np.arange(1,5).reshape(2,2)

# Create no copy only assignment
arr_no_copy=arr

# Create Deep Copy
arr_copy=arr.copy()

# Create shallow copy using View
arr_view=arr.view()

print("Original Array: ",id(arr))
print("Assignment: ",id(arr_no_copy))
print("Deep Copy: ",id(arr_copy))
print("Shallow Copy(View): ",id(arr_view))
```

```python
# Update the values of original array
arr[1]=[99,89]

# Check updated original array
print("Original Array:\n", arr)

# Check values of array view
print("View Array:\n", arr_view)

# Check values of array copy
print("Copied Array:\n", arr_copy)
```

### Date Features 

```python
# Import important library 
import pandas as pd

# Create a DataFrame with date column 
dates = pd.Series(pd.date_range('2024-10-01 00:00:00', freq='19H30min',  periods=10)) 
data = pd.DataFrame(dict(date=dates)) 

# Show the top-5 records 
data.head() 
```

```python
# Extract year, month, day, hour, and minutes 
data['year'] = data.date.dt.year 
data['month'] = data.date.dt.month 
data['day'] = data.date.dt.day 
data['hour'] = data.date.dt.hour 
data['minutes'] = data.date.dt.minute 

# Show the top-5 records 
data.head() 
```

```python
# Extract day_of_week, day_of_year, and quarter 
data['day_of_week'] = data.date.dt.dayofweek 
data['day_of_year'] = data.date.dt.dayofyear 
data['quarter'] = data.date.dt.quarter 

# Show the top-5 records 
data.head() 
```
